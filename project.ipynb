```python
from numpy import inf, sqrt

#def leave_one_out_cross_validation(data, current_set, feature_to_add):
def mymom(data, current_set, feature_to_add):
    features_to_use = list(current_set) + [feature_to_add]
    number_correctly_classified = 0
    
    for i in range(data.shape[0]):
        object_to_classify = data.iloc[i, features_to_use].values
        label_object_to_classify = data.iloc[i, 0]
        
        nearest_neighbor_distance = inf
        #nearest_neighbor_location = inf
        nearest_neighbor_label = None
        
        for k in range(data.shape[0]):
            if k != i:
                candidate_object = data.iloc[k, features_to_use].values
                distance = sqrt(sum((object_to_classify - candidate_object) ** 2))
                
                if distance < nearest_neighbor_distance:
                    nearest_neighbor_distance = distance
                    nearest_neighbor_label = data.iloc[k, 0]
        
        #print(f'Ask if {i} is nearest neighbors with ', {k}, 'which is ', nearest_neighbor_label)
        if label_object_to_classify == nearest_neighbor_label:
            number_correctly_classified += 1
    
    accuracy = number_correctly_classified / data.shape[0]
    return accuracy
```

```python
def feature_search_demo(data):
    current_set_of_features = [] #Initialize an empty set 
    num_features = data.shape[1] - 1 

    best_overall_accuracy = -1.0
    best_overall_set = []
 
    for i in range(1, num_features + 1):
        print(f'On the {i}th level of the search tree')
        feature_to_add_at_this_level = None
        best_so_far_accuracy = -1.0
        
        for k in range(1, num_features + 1):
            if k not in current_set_of_features: #Only consider adding, if not already added.
                print(f'--Consider adding the {k}\'th feature')  
                accuracy = leave_one_out_cross_validation(data, current_set_of_features,k);

                if accuracy > best_so_far_accuracy:
                  best_so_far_accuracy = accuracy
                  feature_to_add_at_this_level = k
                
        current_set_of_features.append(feature_to_add_at_this_level)
        print(f'On level {i} i added feature {feature_to_add_at_this_level} to current set')
        print(f"Current set: {current_set_of_features}, best accuracy: {best_so_far_accuracy:.4f}")

        if best_so_far_accuracy > best_overall_accuracy:
            best_overall_accuracy = best_so_far_accuracy
            best_overall_set = current_set_of_features.copy()

    print(f"Best overall set: {best_overall_set}, best overall accuracy: {best_overall_accuracy:.4f}")
    return current_set_of_features
```

```python
results = feature_search_demo(data)
print(results)
```
